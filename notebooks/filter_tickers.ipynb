```python
# use pandas to read the csv file
import pandas as pd

from config.config import GET_FINANCIAL_DATA

file_path = path = GET_FINANCIAL_DATA["INPUT_CSV_PATH"] 


```

```python
#There are different companies that have the same tickers, we need to filter them out.We cant just dropping the duplicate ones, we need to choose the duplicates and instances we want to keep.

#group by ticker and count the number of unique companies for each ticker, then filter out the tickers that have more than 1 unique company, and also keep the company names for those tickers in a new column called "companies"
df = pd.read_csv(file_path)
ticker_counts_df= df.dropna(subset=['ticker', 'company']).groupby('ticker').agg(company_count=('company', 'nunique'),companies = ('company', lambda x: ', '.join(x.unique()))).reset_index()

ticker_counts_df = ticker_counts_df[ticker_counts_df['company_count'] > 1]


```
