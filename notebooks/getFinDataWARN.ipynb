Cleaned data, removed company that has layoff size below 5

```python


import yfinance as yf
import pandas as pd

df = pd.read_csv(r"../src/final_layoffs.csv")

df = df[df['layoff_size'] >= 5]

df.to_csv(r"layoffSizeFiltered.csv", index=False)
```

```python
import yfinance as yf
import pandas as pd

df = pd.read_csv(r"layoffSizeFiltered.csv")

tickers = []

for company in df['Company_name']:
    results = yf.Search(company).quotes
    if results:
        search_result = results[0]
        tickers.append(search_result.get('symbol'))
    else:
        tickers.append(None)

df['Ticker'] = tickers
df.to_csv(r"layoffs_with_tickers_WARN.csv", index=False)
```

```python
import yfinance as yf
import pandas as pd

df = pd.read_csv(r"layoffs_with_tickers_WARN.csv")
df.dropna(subset=['Ticker'], inplace=True)
df.to_csv(r"cleaned_layoffs_with_tickers_WARN.csv", index=False)
```
