Cleaned data, removed company that has layoff size below 5 from the final_layoff.csv

```python
import pandas as pd
from layoff import config

df = pd.read_csv(config.DATA_CLEANED / "cleaned_layoffs.csv")

df = df[df['layoff_size'] >= 5]

df.to_csv(config.GET_FINANCIAL_DATA_COMBINED["LAYOFF_FILTERED_COMBINED_CSV_PATH"], index=False)
```

```python
import yfinance as yf
import pandas as pd

df = pd.read_csv(config.GET_FINANCIAL_DATA_COMBINED["LAYOFF_FILTERED_COMBINED_CSV_PATH"])

tickers = []

for company in df['Company_name']:
    results = yf.Search(company).quotes
    if results:
        search_result = results[0]
        tickers.append(search_result.get('symbol'))
    else:
        tickers.append(None)

df['Ticker'] = tickers
df.to_csv(config.GET_FINANCIAL_DATA_COMBINED["LAYOFFS_WITH_TICKERS_COMBINED_CSV_PATH"], index=False)
```
