```python
import pandas as pd
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

layoffs = PROJECT_ROOT / "data" / "processed" / "layoffs_with_tickers_cleaned.csv"

df = pd.read_csv(layoffs)
print("Total unique tickers in layoffs data:", df['Ticker'].nunique())

nasdaq = PROJECT_ROOT / "data" / "raw" / "nasdaq-listed.csv"
nasdaq_df = pd.read_csv(nasdaq)
print("Total unique symbols in NASDAQ data:", nasdaq_df['Symbol'].nunique())

#cross checking the tickers in the cleaned layoffs data with the nasdaq-listed tickers
cross_check = df['Ticker'].isin(nasdaq_df['Symbol'])
print("Number of tickers in layoffs data that are in NASDAQ data:", cross_check.sum())


laid_off_tickers = set(df["Ticker"])

no_layoff = nasdaq_df[~nasdaq_df["Symbol"].isin(laid_off_tickers)]

print("Total unique tickers in NASDAQ data that are NOT in layoffs data:", no_layoff['Symbol'].nunique())

print(f"No-layoff pool: {no_layoff.shape}, companies={no_layoff['company'].nunique()}")
no_layoff.to_csv(PROJECT_ROOT / "data" / "processed" / "no_layoff_tickers.csv", index=False)

no_layoff.head()
```
