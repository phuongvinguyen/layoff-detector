Given the pdf we pulled from https://airtable.com/app1PaujS9zxVGUZ4/shroKsHx3SdYYOzeh/tblleV7Pnb6AcPCYL?viewControls=on, we want to extract the information as a df and later safe it as a csv for easy accessing.

```python
# imports
import pandas as pd
from tabula.io import read_pdf
from config import PDF_TO_CSV
# Pandas option
pd.set_option('display.max_columns', None)
# Path
PDF_PATH = PDF_TO_CSV["PDF_INPUT_PATH"]
dfs = read_pdf(PDF_PATH, pages='all', multiple_tables=True)
# Merge dfs into a single df
df = pd.concat(dfs)
# Present df for first check
df
```

```python
df = df.iloc[:-1] # Remove the last row (only containing information on total employees fired and total money raised)
df
df.to_csv(PDF_TO_CSV["CSV_OUTPUT_PATH"], index=False)
```
