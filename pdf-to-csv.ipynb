```python
# imports
import pandas as pd
from tabula.io import read_pdf
import config.py
# Pandas option
pd.set_option('display.max_columns', None)
# Path
PDF_PATH = config.PDF_PATH
dfs = read_pdf(PDF_PATH, pages='all', multiple_tables=True)
# Merge dfs into a single df
df = pd.concat(dfs)
# Present df for first check
df
```
